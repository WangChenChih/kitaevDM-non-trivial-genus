```python
import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib
from GeoIllu import GeoIlluLib as gil
```

```python
illu = gil.LatticeCreation()
```

# Normal Honeycomb

```python
#pair_x, pair_y, dual_lattice_x, dual_lattice_y, num_pair = illu.HoneycombLatticeEdge(width=int(7), height=int(13), latconst=1, vertical_length=0.3)
pair_x, pair_y, dual_lattice_x, dual_lattice_y, num_pair = \
illu.HoneycombLatticeEdge(width=15, height=25, latconst=1)

plt.figure()
for i in range(num_pair):
    plt.plot(pair_x[i,:], pair_y[i,:], 'b-', linewidth=0.5)
#plt.scatter(dual_x, dual_y, c='k', s=15)    # plot dual lattice points
plt.gca().set_aspect('equal', adjustable='box')
plt.gca().set_axis_off()
plt.tight_layout()
plt.savefig('example.svg', format='svg')
```

```python
pair_x, pair_y, leng = illu.HoneycombLatticeEdge_Center(layer=4, latconst=1)

plt.figure()
for i in range(leng):
    plt.plot(pair_x[i,:], pair_y[i,:], 'b-', linewidth=0.5)
plt.gca().set_aspect('equal', adjustable='box')
plt.gca().set_axis_off()
plt.tight_layout()
plt.savefig('example-center.svg', format='svg')
```

```python
pair_x, pair_y, leng = illu.HoneycombLatticeEdge_Center(layer=1, latconst=1)

plt.figure()
for i in range(leng):
    plt.plot(pair_x[i,:], pair_y[i,:], 'k-', linewidth=1.5)

tri_up_x = [-np.sqrt(3)/2,np.sqrt(3)/2,0,-np.sqrt(3)/2]
tri_up_y = [0.5, 0.5,1,0.5]
plt.plot(tri_up_x, tri_up_y,'b--', linewidth=2)
tri_down_x = [-np.sqrt(3)/2,np.sqrt(3)/2,0,-np.sqrt(3)/2]
tri_down_y = [-0.5, -0.5,-1,-0.5]
plt.plot(tri_down_x, tri_down_y,'r--', linewidth=2)
plt.gca().set_aspect('equal', adjustable='box')
plt.gca().set_axis_off()
plt.tight_layout()
plt.savefig('example.svg', format='svg')
```

```python
pair_x, pair_y, leng = illu.HoneycombLatticeEdge_Center(layer=1, latconst=1)

plt.figure()
for i in range(leng):
    plt.plot(pair_x[i,:], pair_y[i,:], 'k-', linewidth=2)

tri_x = [-np.sqrt(3)/2,np.sqrt(3)/2,0,-np.sqrt(3)/2]
tri_y = [0.5, 0.5,1,0.5]
# plot six triangles corresponding to six corners
for j in range(6):
    angle = j * math.pi / 3
    rot_matrix = np.array([[math.cos(angle), -math.sin(angle)],
                           [math.sin(angle),  math.cos(angle)]])
    rotated_tri = rot_matrix @ np.array([tri_x, tri_y])
    plt.plot(rotated_tri[0,:], rotated_tri[1,:],'--', linewidth=1.5)

plt.gca().set_aspect('equal', adjustable='box')
plt.gca().set_axis_off()
plt.tight_layout()
plt.savefig('example.svg', format='svg')
```
